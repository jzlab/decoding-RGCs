```python
import numpy as np
import h5py
import shutil
import matplotlib.pyplot as plt
```

```python
shutil.copy("learned_mosaic_weights.h5", "copy.h5")
```

```python
results = h5py.File('copy.h5', 'r')
epochs = "0002"
```

```python
fig, ax = plt.subplots(1, 5, figsize=(20, 4))

ax[0].imshow(np.array(results[f"epoch_{epoch}"]["summed_subunit_kernel"]).squeeze(0), cmap="gray")
ax[0].set_title("Summed Subunit Kernel")

for i in range(4):
    ax[i+1].imshow(np.array(results[f"epoch_{epoch}"]["functional_subunit_kernels"][i]).squeeze(0), cmap="gray")
    ax[i+1].set_title(f"Functional Subunit Kernel {i + 1}")

for a in ax:
    a.axis("off")

plt.show()
```
